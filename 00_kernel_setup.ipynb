## Other Systems

There is an envirnment.yml file in the repository that can be used to create a Conda environment on other systems that should work to run these notebooks. 

Generally, you will want to run a command like the one below in a terminal on your system:

     conda env create --prefix deep_learning-pt_lightning -f environment.yml

Then to make the kernel definition file:

     conda activate deep_learning-pt_lightning
     python -m ipykernel install --user --name="transfer-learning" --display-name="Transfer Learning"

If you want, you can add the icon by putting the file [logo-64x64.png](images/logo-64x64.png) in your kernel folder, typically at `~/.local/share/jupyter/kernels/transfer-learning`.
